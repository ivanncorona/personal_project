La identificación de faltantes en auditoría es de suma importancia, es una técnica simple pero útil para saber si hay indicios de manipulación deliberada de datos físicos o electrónicos: pólizas, cheques, facturas, etc. Este notebook demuestra el poder y simpleza de python para encontrar faltantes.

```python
import pandas as pd
import numpy as np
```

```python
# ejemplo simple
valores = [1,2,3,5,6,7,9,10]

df = pd.DataFrame(valores)
df.rename(columns={0: 'Valor'}, inplace=True)

min_valor = df['Valor'].min()
max_valor = df['Valor'].max()
todos_valores = set(range(min_valor, max_valor + 1))

# Encontrar los valores faltantes
valores_faltantes = todos_valores - set(df['Valor'])

# Mostrar los valores faltantes
valores_faltantes
```

```python
# importar datos
data = pd.read_excel(r'data\Prueba Benford ejemplo - copia.xls')

# crear un DataFrame
df = pd.DataFrame(data)

# ya que al importarla serie 'cheque ' contiene un espacio la sig línea lo arregla 
df.rename(columns={'cheque ': 'cheque'}, inplace=True)

df.head()
```

```python
# generar una serie con todos los valores entre el mínimo y el máximo de 'cheque'
min_cheque = df['cheque'].min()
max_cheque = df['cheque'].max()
todos_cheques = set(range(min_cheque, max_cheque + 1))

# encontrar los valores faltantes
valores_faltantes = todos_cheques - (set(df['cheque']) )

# ordenar los valores faltantes de menor a mayor
valores_faltantes_ordenados = sorted(valores_faltantes)

# formatear el resultado
resultado = "Valores faltantes: " + ", ".join(map(str, valores_faltantes_ordenados))

# mostrar los valores faltantes ordenados
resultado
```

### Ejemplo utilizando alfanuméricos

```python
# datos
data = {
    'cheque': [
        'AA963', 'AA964', 'AA965', 'AA966', 'AA967', 'AA968', 'AA969', 'AA974', 'AA975', 'AA976',
        'AB156', 'AB157', 'AB161', 'AB162', 'AB163', 'AB164', 'AB165', 'AB166', 'AB167', 'AC3527',
        'AC3528', 'AC3529', 'AC3532', 'AC3533', 'AC3535', 'AC3536'
    ],
    'fecha': [
        '28/12/2005', '28/12/2005', '01/01/2006', '01/01/2006', '03/01/2006', '04/01/2006',
        '04/01/2006', '09/01/2006', '09/01/2006', '09/01/2006', '11/01/2006', '11/01/2006',
        '11/01/2006', '12/01/2006', '12/01/2006', '12/01/2006', '13/01/2006', '13/01/2006',
        '14/01/2006', '16/01/2006', '17/01/2006', '18/01/2006', '18/01/2006', '18/01/2006',
        '20/01/2006', '20/01/2006'
    ],
    'importe': [
        '4,414.99', '4,368.39', '20,000.00', '782', '597.81', '116,631.01', '20,000.00',
        '30,000.00', '25,596.31', '2,152.80', '6,510.76', '7,753.25', '3,122.27', '20,000.00',
        '4,099.75', '1,150.00', '2,746.68', '2,251.82', '527.47', '30,000.00', '590,681.31',
        '14,985.00', '8,656.00', '15,000.00', '3,775.08', '30,000.00'
    ]
}

# Crear un DataFrame
df = pd.DataFrame(data)
df.head(10)
```

```python
# identificar las series de cheques únicas
series_cheques = df['cheque'].str.extract(r'([A-Z]+)').squeeze().unique()

# crear un diccionario para almacenar los valores de cheques faltantes por serie
valores_faltantes = {}

# función para encontrar los valores faltantes en una serie de cheques
def encontrar_faltantes(serie_cheque):
    numeros = df[df['cheque'].str.startswith(serie_cheque)]['cheque'].str.extract(r'(\d+)').astype(int)
    numeros = numeros.squeeze().tolist()
    valores_faltantes = [num for num in range(min(numeros), max(numeros) + 1) if num not in numeros]
    return valores_faltantes

# encontrar los valores faltantes en cada serie de cheques
for serie in series_cheques:
    valores_faltantes[serie] = encontrar_faltantes(serie)

# mostrar los valores faltantes
for serie, faltantes in valores_faltantes.items():
    if faltantes:
        print(f'Valores faltantes en serie {serie}: {", ".join([f"{serie}{f}" for f in faltantes])}')
    else:
        print(f'No hay valores faltantes en serie {serie}')
```
