Ya trabajando en una institución financiera, regularme codificaba para agilizar mi trabajo, gracias a una versión acotada pero funcional de Anaconda, la cual era necesario descargar para estudiar un curso introductorio a Python en la plataforma Udemy. Fue un día cuando asistí a una reunión, donde se debatía la forma de entregar la información a un regulador. Todos estaban ansiosos, la fecha de entrega era muy próxima y apenas se había recopilado la información, el problema en ese momento y la razón por la que todos estaban nerviosos, era el tiempo de entrega del área de TI puesto que la “complejidad” era la de unir cerca de 120 archivos Excel, los cuales tenían la misma estructura y correspondían, como seguramente ya dedujiste, a datos diarios de 4 meses. 

Durante la llamada me puse en contacto con el organizador, quien era mi cliente al pertenecer a una unidad de negocio que yo atendía, le dije que podía intentar algo y le pedí acceso a la carpeta. La llamada se extendió por razones que no recuerdo, pero se argumentaba de la complejidad de la situación y en cuanto tiempo TI atendería la solicitud. Momentos antes de terminar la llamada yo ya tenía este código en el notebook de Anaconda:

```
ruta = r'D:\ruta a los 120 archivos'
archivos = [f for f in os.listdir(ruta) if f.endswith('.csv')]

dataframes = []

for archivo in archivos:
    df = pd.read_csv(os.path.join(ruta, archivo), encoding='latin1')
    df['origen'] = os.path.splitext(archivo)[0]
    dataframes.append(df)

final = pd.concat(dataframes, ignore_index=True)
final
```

Tomé la palabra, traté de no hacer mucho ruido al mostrar mi pantalla, mostrando el dataframe consolidado de los 120 archivos que se traducían en al menos 2 millones de registros, sin embargo, mi pantalla causó conmoción… y entiendo por qué, recuerdo bien que sume una “columna” de la tabla consolidada y por pura coincidencia lo operativo difería en menos de 1% contra lo contable, monetariamente la diferencia no era material al momento de presentar las cifras en miles de millones de pesos, por lo que podíamos utilizar estos datos para iniciar con el análisis y posterior entrega de información, sobra decir que muchas personas en la llamada estaban sorprendidas, ahora que lo pienso para una persona que no ha tenido contacto con código, con recursión, un lenguaje de programación resulta esotérico, casi mágico.
