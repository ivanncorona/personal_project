### Introducción

La razón de tener este espacio en GitHub es bien simple: desmitificar el uso de código y técnicas avanzadas de análisis de datos, métodos estadísticos, incluso machine learning, para fines de auditoría, evaluación de riesgos, clasificación, forecast de datos, etc.

Soy un contador con amplia experiencia en auditoría, control interno y análisis de datos. Desde niño siempre fui una persona con mucha curiosidad. Mi primer contacto con un lenguaje de programación (aunque muchos debaten si lo es o no) fue con SQL vía MS Access.

Gracias al conocimiento que adquirí en SQL tuve la oportunidad de participar en un proyecto como Software engineer, donde el Banco donde participé utilizaba ampliamente R para programar algunas revisiones. Por esta situación, decidí empezar a estudiar las estructuras básicas de R,loops, condiciones, funciones, etc., pero el énfasis siempre fue la estructura conocida como _dataframe_, la cual representa datos en formato tabular, tal como una tabla de SQL o datos en una hoja de Excel.

No tardé mucho en darme cuenta de que, por cada libro, página de internet o curso que consultaba de R había por lo menos 10 de Python, incluso mucho más especializados en la profesión contable, enfocados en finanzas o incluso en la evaluación de riesgos, por ello decidí aprender Python de forma totalmente autodidacta, primero reproduciendo lo que regularmente hacía analizando datos con Excel, e incluso reproduciendo las funcionalidades que ofrecían analíticos comerciales como IDEA o ACL.

La primera vez que en verdad, de forma consciente, vi el poder de un lenguaje de programación y lo que puede aportar fue la siguiente. Ya trabajando en una institución financiera, regularme codificaba para agilizar mi trabajo, gracias a una versión acotada pero funcional de Anaconda. Un día cuando asistí a una reunión, donde se debatía la forma de entregar la información a un regulador todos estaban ansiosos, la fecha de entrega era muy próxima y apenas se había recopilado la información operativa, "el problema" en ese momento y la razón por la que todos estaban nerviosos, era el tiempo de entrega del área de TI, puesto que la supuesta complejidad era la de unir cerca de 120 archivos Excel, TI se había comprometido en entregar en un lapso no mayor a 24 hrs...

Durante la llamada me puse en contacto con el organizador, quien era mi cliente al pertenecer a una unidad de negocio que yo atendía, le dije que podía intentar algo y le pedí acceso a la carpeta. Momentos antes de terminar la llamada yo ya tenía algo parecido a este código en el notebook de Anaconda:

```
ruta = r'D:\ruta a los 120 archivos'
archivos = [f for f in os.listdir(ruta) if f.endswith('.csv')]

dataframes = []

for archivo in archivos:
    df = pd.read_csv(os.path.join(ruta, archivo), encoding='latin1')
    df['origen'] = os.path.splitext(archivo)[0]
    dataframes.append(df)

final = pd.concat(dataframes, ignore_index=True)
final
```

Tomé la palabra, mostré el resultado en pantalla, el dataframe consolidado con los 120 archivos se traducían en al menos 2 millones de registros, resultó una conmoción… lo operativo difería en menos de 1% contra lo contable, monetariamente la diferencia no era material al momento de presentar las cifras en miles de millones de pesos, por lo que podíamos utilizar estos datos para iniciar con la limpieza, el análisis y posterior entrega de información, sobra decir que muchas personas en la llamada estaban sorprendidas, ahora entiendo porque: para una persona que jamás ha tenido contacto con código, con recursión, un lenguaje de programación resulta esotérico y sus resultados se presentan como algo casi mágico.

El código tiene 8-9 líneas, pero si lo piensas, tan solo con eso estás importando, como en este ejemplo, millones de registros, los cuales están listos para hacer un Exploratory Data Analysis, limpieza, etc., etc., ese es el poder y la simplicidad de Python.

Desde mi humilde experiencia como contador:

- Relacionado con el punto anterior, la codificación/programación es una labor intelectual de creación, casi como la pintura, escultura u otra forma de elaborar algo. Es bastante satisfactorio crear algo que funciona, que resuelve un problema y/o que responde una pregunta.
- Con disciplina puedes aprender cualquier cosa, desde mi perspectiva puedes ser tan bueno en análisis de datos, casi como cualquier profesional en ciencia de datos [Nota], sin tener toda la instrucción académica.

Lo que nos distingue de los especialistas en datos es el _**fit**_ de los modelos, ellos ajustan los modelos de una forma que para la amplia mayoría de contadores les resulta (tan solo en teoría) imposible hacer, sin embargo, con la formación necesaria es totalmente posible.
